## 1. The five words you must know

- **Repository (repo):** a project folder tracked by Git.
- **Commit:** a saved checkpoint with a message.
- **Branch:** a separate line of development.
- **Push:** upload local commits to a remote repository such as GitHub.
- **Pull:** download commits from the remote repository.

### Mental model
Think of Git as video-game save points. A commit is a save point; a branch is an alternate path; GitHub is the online copy.

## 4. Create the tiny project
We will build the same small project throughout the lab. Git will track every checkpoint.

```python
from pathlib import Path
import os, subprocess, textwrap

PROJECT = Path("/content/bano-qabil-git-demo")
PROJECT.mkdir(parents=True, exist_ok=True)
os.chdir(PROJECT)

Path("README.md").write_text(
    "# Bano Qabil Agentic AI\n\nThis is my version-control practice project.\n",
    encoding="utf-8"
)

print("Project created at:", PROJECT)
print(Path("README.md").read_text())
```

## 5. Turn the folder into a Git repository
This is the Colab equivalent of clicking **Create a New Repository** in GitHub Desktop.

```python
!git init
!git status
```

## 6. First commit — our first save point
Stage the file, create a commit, and inspect the history.

```python
!git add README.md
!git commit -m "Initial project"
!git log --oneline --decorate --all
```

## 7. Make a change and inspect what Git notices
Before committing, Git can show exactly what changed.

```python
Path("README.md").write_text(
    "# Bano Qabil Agentic AI\n\nThis is my version-control practice project.\n\n## Week 1\nLearned the basic Git workflow.\n",
    encoding="utf-8"
)

!git status
!git diff -- README.md
```

## 8. Second commit
Commit messages should describe the change clearly.

```python
!git add README.md
!git commit -m "Added Week 1 notes"
!git log --oneline --decorate --all
```

## 9. Create a branch
Imagine you want to experiment without changing the official `main` version. This is exactly what a branch is for.

```python
!git switch -c feature-homepage
!git branch
```

## 10. Work safely on the branch
We will add a small homepage file. This change belongs to the feature branch until we merge it.

```python
Path("index.html").write_text(
    "<!doctype html>\n<html>\n<head><title>Bano Qabil</title></head>\n"
    "<body><h1>My Agentic AI Project</h1></body>\n</html>\n",
    encoding="utf-8"
)

!git status
!git add index.html
!git commit -m "Added homepage"
!git log --oneline --decorate --all --graph
```

## 11. Merge the branch into `main`
First return to `main`, then merge the completed feature branch.

```python
!git switch main
!git merge feature-homepage --no-edit
!git log --oneline --decorate --all --graph
!git status
```

## 12. See the complete project history
Git remembers the checkpoints so you can inspect how the project changed over time.

```python
!git log --oneline --decorate --all --graph
print("\nCurrent files:")
for p in sorted(PROJECT.iterdir()):
    if p.name != ".git":
        print(" -", p.name)
```

## 13. What Push and Pull mean

At this point the repository exists only inside this Colab runtime.

- **Push** sends your local commits to a remote GitHub repository.
- **Pull** brings newer commits from that remote repository to your local repository.

GitHub Desktop provides buttons for these operations. In the original Desktop handout, you do not need to type commands.
